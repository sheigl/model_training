```python
from pymongo import MongoClient
```

```python
client = MongoClient("mongodb://server.home:27017/", username="root", password="whatever", authSource='admin')
```

```python
card_collection = client['mtg_json']['cards']
combo_collection = client['commander_spellbook']['variants']
synthetic_collection = client['synthetic_queries']['queries']
```

```python
all_combos = list(combo_collection.find({'status': 'OK'}).limit(5000))
mongo_documents = []
```

```python
card_combo_count = {}
for combo in all_combos:
    cards = combo.get('uses', [])
    for card_info in cards:
        card_name = card_info.get('card', {}).get('name', '')
        if card_name:
            card_combo_count[card_name] = card_combo_count.get(card_name, 0) + 1
```

```python
popular_cards = sorted(card_combo_count.items(), key=lambda x: x[1], reverse=True)[:200]
```

```python
print(popular_cards)
```
